## 3.2 Functions and graphs: the maths you already write in code

A model is a function, so reading f(x) and its graph comes first. You plot each one in Python as you go.

[Read the lesson](https://aicourses.dev/research/math-warm-up/functions-and-graphs)

### A function is a machine

In Python you have written this kind of thing since Module 2:

```python
def fare(km):
    return 2 * km + 3      # 3 to get in, then 2 for each km

print(fare(4))             # 11
```

### A graph is every answer at once

NumPy, from lesson 2.5, makes the table in one go. Give the function a whole array, and `*` and `+` act on every number in it:

```python
import numpy as np
import matplotlib.pyplot as plt

xs = np.arange(0, 6)       # the whole numbers 0 to 5
print(fare(xs))            # [ 3  5  7  9 11 13]

plt.plot(xs, fare(xs), marker="o")
plt.xlabel("distance (km)")
plt.ylabel("fare")
plt.show()
```

### Not every graph is a straight line

To draw a smooth curve, ask NumPy for many inputs close together. `np.linspace(0, 5, 101)` gives 101 evenly spaced numbers from 0 to 5 (0, 0.05, 0.1 and so on), and `**` is Python’s power, so `xs ** 2` squares each one:

```python
xs = np.linspace(0, 5, 101)   # 101 evenly spaced points
plt.plot(xs, 2 * xs + 3, label="fare: 2x + 3")
plt.plot(xs, xs ** 2, label="area: x squared")
plt.legend()
plt.show()
```

## 3.3 Reading the notation: Σ, subscripts and vectors

Papers and later lessons write loops as Σ and arrays as xᵢ. Translating each symbol into a for loop removes most of the fear of equations.

[Read the lesson](https://aicourses.dev/research/math-warm-up/sigma-and-subscripts)

### Σ: a for loop in one symbol

Now look at the same thing as code. Each part of the symbol has a matching part in the loop:

```python
x = [3, 5, 4, 6, 2]    # steps each day, in thousands
n = len(x)

total = 0
for i in range(n):     # Python's i runs 0, 1, 2, 3, 4
    total = total + x[i]
print(total)           # 20
```

### The same thing in NumPy

A loop shows what Σ means, but NumPy, from lesson 2.5, does it in one call and much faster on long lists:

```python
import numpy as np

x = np.array([3, 5, 4, 6, 2])
print(x.sum())          # 20
print(x[1:4].sum())     # 15, days 2 to 4
print(x.sum() / len(x)) # 4.0, the average day
```

### Adding up products

Here $q_i p_i$ means $q_i$ times $p_i$: each line of the receipt, how many times the price. Σ then adds up the lines. NumPy multiplies two arrays item by item with `*`, so the whole receipt is one line:

```python
q = np.array([3, 2, 1])        # how many of each thing
p = np.array([0.5, 2.0, 1.5])  # the price of each thing
print(q * p)                   # [1.5 4.  1.5]
print(np.sum(q * p))           # 7.0
```

## 3.4 Exponents and the number e

Repeated multiplication makes numbers grow very fast. The number e, about 2.718, is the base models use, and later a model uses it to turn raw scores into probabilities.

[Read the lesson](https://aicourses.dev/research/math-warm-up/exponents-and-e)

### Powers: multiplying again and again

In Python, the power sign is `**`:

```python
thickness = 0.1                   # one sheet, in mm
print(thickness * 2 ** 10)        # 102.4, in mm
print(thickness * 2 ** 42 / 1e6)  # 439804.65111040004, in km
```

### The number e, from a bank account

Here $n$ is how many times a year the interest is paid, and $1 + \frac{1}{n}$ is what each payment multiplies your money by. For $n = 2$ that is $1.5^2 = 2.25$. Paying more often does not make you endlessly rich. The answer creeps up towards one particular number:

```python
for n in [1, 2, 12, 365, 1_000_000]:
    print(n, (1 + 1 / n) ** n)
# 1 2.0
# 2 2.25
# 12 2.613035290224676
# 365 2.7145674820219727
# 1000000 2.7182804690957534
```

### Why models love e

That pair is exactly what you need to turn a model’s raw scores, which can be any number, into chances, which must be positive and add up to 1. Raise e to each score, then divide each by the total:

```python
import numpy as np

scores = np.array([2.0, 1.0, 0.0])  # raw scores for 3 choices
ups = np.exp(scores)                # always positive
print(ups)              # [7.3890561  2.71828183 1.        ]
print(ups / ups.sum())  # [0.66524096 0.24472847 0.09003057]
```

## 3.5 Logarithms: the power you need

A log answers “what power gives this number?”, and it turns multiplying into adding. The loss every language model in this course trains on is a log, and log scales let you draw numbers from 1 to a billion on one chart.

[Read the lesson](https://aicourses.dev/research/math-warm-up/logarithms)

### A log asks “what power?”

Now the paper. The Moon is about 384,400 km away, which is $3.844 \times 10^{11}$ mm, or about $3.844 \times 10^{12}$ sheets of 0.1 mm. Each fold doubles the number of sheets in the stack, so the number of folds is how many doublings reach that many sheets, its log base 2:

```python
import numpy as np

moon_mm = 384_400 * 1_000_000   # 384,400 km, in mm
print(np.log2(moon_mm / 0.1))   # 41.80574547476001
```

### Logs turn multiplying into adding

Here $a$ and $b$ are any two positive numbers, and the rule works in any base, as long as you use the same one throughout. With $a = 100$ and $b = 1000$ in base 10: $\log_{10}(100{,}000) = 5$, and $\log_{10} 100 + \log_{10} 1000 = 2 + 3 = 5$.

```python
a, b = 100, 1000
print(np.log10(a * b), np.log10(a) + np.log10(b))   # 5.0 5.0
```

```python
p = np.full(400, 0.1)      # 400 chances, each 0.1
print(np.prod(p))          # 0.0
print(np.sum(np.log(p)))   # -921.0340371976183
```

## 3.6 Probability basics: chance, distributions and sampling

A language model outputs a probability for every possible next token. Distributions and sampling are what you need to read that output and generate from it.

[Read the lesson](https://aicourses.dev/research/math-warm-up/probability-basics)

### Sampling: letting chance pick

To **sample** from a distribution is to pick one outcome at random, with each outcome as likely as its probability says. Closing your eyes and picking a sweet is sampling. NumPy can do it for you:

```python
import numpy as np

colours = ["red", "green", "yellow"]
probs = [0.5, 0.3, 0.2]       # 5, 3 and 2 sweets out of 10

rng = np.random.default_rng(seed=0)
print(rng.choice(colours, size=10, p=probs))
# ['green' 'red' 'red' 'red' 'yellow' 'yellow' 'green'
#  'green' 'green' 'yellow']
```

### Many samples settle down

Draw more, and the shares settle close to the true chances. Continue with the same `rng`:

```python
draws = rng.choice(colours, size=10_000, p=probs)
for c in colours:
    print(c, np.mean(draws == c))
# red 0.4988
# green 0.2991
# yellow 0.2021
```

## 3.7 Chance given what you already know: conditional probability

A language model’s guess is the probability of the next token, given the text so far. Multiplying these chances gives the chance of a whole sentence, and that is how Module 9 scores its first model.

[Read the lesson](https://aicourses.dev/research/math-warm-up/conditional-probability)

### Counting next words

A language model’s question is a conditional probability: $P(\text{next token} \mid \text{text so far})$. You can estimate a simple version by counting, just like the diary. Take a tiny text and look at which word follows which:

```python
text = "the cat sat on the mat . the cat sat . the dog ran ."
words = text.split()
pairs = list(zip(words, words[1:]))    # each word with the next
print(pairs[:3])
# [('the', 'cat'), ('cat', 'sat'), ('sat', 'on')]

after_the = [b for a, b in pairs if a == "the"]
print(after_the)       # ['cat', 'mat', 'cat', 'dog']
print(after_the.count("cat") / len(after_the))    # 0.5
```

### The chance of a whole sentence

This version looks back only one word at a time, to keep the counting simple. A real model looks at all of the text so far.

```python
import numpy as np

def p_next(word, prev):
    after = [b for a, b in pairs if a == prev]
    return after.count(word) / len(after)

chances = [p_next("cat", "the"), p_next("sat", "cat"),
           p_next("on", "sat")]
print(chances)                   # [0.5, 1.0, 0.5]
print(np.prod(chances))          # 0.25
print(np.sum(np.log(chances)))   # -1.3862943611198906
```

## 3.8 Expected value: the average you get in the long run

If you repeat something chancy many times, the average result is its expected value: each outcome times its chance, added up. Losses, eval scores and RL rewards are all built on expected values.

[Read the lesson](https://aicourses.dev/research/math-warm-up/expected-value)

### Each outcome times its chance

The formula is the same multiply-and-add pattern as the shopping bill in lesson 3.3, so NumPy computes it the same way:

```python
import numpy as np

values = np.array([1, 2, 3, 4, 5, 6])
probs = np.full(6, 1 / 6)          # a fair die
print(np.sum(values * probs))      # 3.5

winnings = np.array([0, 10])       # nothing, or 10 for a six
probs = np.array([5 / 6, 1 / 6])
print(np.sum(winnings * probs) - 2)  # -0.3333333333333335
```

### Watching the average settle

Expected value is a prediction about long runs, so test it with one. Roll a die 10,000 times and track the **running average**, the average of all rolls so far, after each one:

```python
rng = np.random.default_rng(seed=0)
rolls = rng.integers(1, 7, size=10_000)    # whole numbers 1 to 6
running = np.cumsum(rolls) / np.arange(1, 10_001)
print(running[[9, 99, 999, 9999]])
# [2.8    3.62   3.6    3.5045]
```

### Where expected values hide in AI

- **Losses.** The loss from lesson 3.5 is measured on one example. Training reports the average loss over many examples, an estimate of the expected loss on text like the training data.
- **Eval scores.** Score a model 1 for each right answer and 0 for each wrong one. The expected value of that score is $1 \times P(\text{right}) + 0 \times P(\text{wrong}) = P(\text{right})$, so an accuracy, the average of those scores, estimates the chance the model gets a question like these right.
- **Rewards.** In reinforcement learning, which you met in Module 0, a model gets a reward for each attempt, and training tries to raise its expected reward, the average over many attempts.

```python
right = np.array([1, 0, 1, 1, 0])   # 1 if the model got it right
print(right.mean())                 # 0.6
```

## 3.9 Averages, spread and noise: mean, variance and standard deviation

Every experiment is noisy. Mean and spread are how you tell a real improvement from luck, from training curves to evaluations.

[Read the lesson](https://aicourses.dev/research/math-warm-up/averages-and-noise)

### Spread: variance and standard deviation

The variance is in “minutes squared”, which is hard to picture. Its square root, $\sigma$, the **standard deviation**, brings it back to minutes: $\sqrt{130} \approx 11.4$ minutes. For bus A the same steps give a variance of 0.4 and a standard deviation of about 0.63 minutes. Bus B’s spread is about 18 times bigger.

```python
import numpy as np

bus_a = np.array([19, 20, 21, 20, 20])   # minutes, five trips
bus_b = np.array([5, 35, 20, 10, 30])

dev = bus_b - bus_b.mean()   # how far each trip is from the mean
print(dev)                   # [-15.  15.   0. -10.  10.]
print(np.mean(dev ** 2))     # 130.0, the variance

print(bus_a.std(), bus_b.std())
# 0.6324555320336759 11.40175425099138
```

### Noise: the same thing, measured twice

You can measure the noise by simulating many tests. `rng.random` gives random numbers between 0 and 1, so each one is below 0.7 with a chance of 0.7, a right answer:

```python
rng = np.random.default_rng(seed=0)
right = rng.random((1000, 100)) < 0.7   # 1,000 tests of 100
scores = right.mean(axis=1)             # one score per test
print(scores[:5])         # [0.63 0.64 0.61 0.69 0.74]
print(scores.std())       # 0.04503380841101494
print(np.mean(scores >= 0.72))          # 0.375
```

### Shrinking the noise

More questions mean less noise. Run the same simulation with 10,000 questions per test:

```python
right = rng.random((1000, 10_000)) < 0.7
scores = right.mean(axis=1)
print(scores.std())                  # 0.004488878567081094
print(np.mean(scores >= 0.72))       # 0.0
```

## 3.10 Slopes and rates of change: the idea behind gradients

Training walks downhill along the slope. Seeing a slope as how much the output moves when you nudge the input prepares Module 7’s gradients.

[Read the lesson](https://aicourses.dev/research/math-warm-up/slopes-and-change)

### Nudging in code

The formula is one line of Python:

```python
def f(x):
    return x ** 2

def slope(f, x, h):
    return (f(x + h) - f(x)) / h      # rise over run

for h in [1, 0.1, 0.01, 0.001]:
    print(h, slope(f, 3, h))
# 1 7.0
# 0.1 6.100000000000012
# 0.01 6.009999999999849
# 0.001 6.000999999999479
```

### Why training cares: walking downhill

You do not need to know where the bottom is. Stand at $w = 1$, measure the slope, and step the other way. A negative slope means the loss falls as $w$ grows, so you increase $w$:

```python
def loss(w):
    return (w - 4) ** 2 + 1      # lowest when w is 4

w = 1.0
print(loss(w), slope(loss, w, 0.001))   # 10.0 -5.998999999999199
w = w - 0.1 * slope(loss, w, 0.001)     # step against the slope
print(w, loss(w))   # 1.59989999999992 6.760480010000384
```
