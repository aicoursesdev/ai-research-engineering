### Lists and dicts: holding things

A **list** holds items in order, inside square brackets. Training code uses lists all the time, for example to record the loss after each round:

````python
losses = [2.31, 1.87, 1.52, 1.40, 1.38]   # one loss per round

print(losses[0])      # the first item: counting starts at 0
print(losses[-1])     # the last item
print(losses[1:3])    # items 1 and 2: the end is left out
print(len(losses))    # how many items

losses.append(1.37)   # add one to the end
print(min(losses))
````

A **dict**, short for dictionary, holds values under names called **keys**, inside curly braces. ML code uses dicts for settings and for logs:

````python
config = {"lr": 0.01, "batch_size": 32, "model": "tiny"}

print(config["lr"])                # look up by key
config["epochs"] = 3               # add a new key
print(config.get("dropout", 0.0))  # a default if it is missing

for key, value in config.items():
    print(key, "=", value)
````

### Loops and comprehensions: working through things

Python loops over the items themselves, not over positions. Two helpers come up constantly. `enumerate` gives you the position and the item together, and `zip` walks through two lists side by side:

````python
losses = [2.31, 1.87, 1.52, 1.40, 1.38]

for step, loss in enumerate(losses):
    if step % 2 == 0:
        print(f"step {step}: loss {loss:.1f}")

names = ["hunger", "tiredness", "money"]
weights = [2.0, 1.5, 1.0]
for name, w in zip(names, weights):
    print(name, w)
````

Building a new list from an old one is so common that Python has a short form for it, the **comprehension**:

````python
losses = [2.31, 1.87, 1.52, 1.40, 1.38]

# the long way
rounded = []
for loss in losses:
    rounded.append(round(loss))
print(rounded)

# the same thing as a comprehension
print([round(loss) for loss in losses])
print([loss for loss in losses if loss < 1.5])

names = ["hunger", "tiredness", "money"]
weights = [2.0, 1.5, 1.0]
print({n: w for n, w in zip(names, weights)})
````

### Functions and keyword arguments: settings with defaults

A function is defined with `def`. ML code is full of functions with many settings, so it gives most of them **default values** and lets you change any one by name, as a **keyword argument**:

````python
def predict(distance, per_km=3.0, start=10.0):
    """Delivery time in minutes, as in lesson 0.4."""
    return per_km * distance + start

print(predict(4))                      # uses the defaults
print(predict(4, start=5.0))           # change one, by name
print(predict(distance=4, per_km=2.0, start=15.0))
````

### Classes: a model with its knobs

A **class** bundles data with the functions that use it, which are called **methods**. Each thing you make from a class is an **object**. Models are almost always classes: the knobs are stored on the object, and methods such as `predict` use them. Here is the pizza model from lessons 0.4 and 0.5 as a class:

````python
class DeliveryModel:
    def __init__(self, per_km, start):
        self.per_km = per_km      # the model's two knobs
        self.start = start

    def predict(self, distance):
        return self.per_km * distance + self.start

    def loss(self, distances, times):
        misses = [abs(self.predict(d) - t)
                  for d, t in zip(distances, times)]
        return sum(misses) / len(misses)

model = DeliveryModel(per_km=2.0, start=15.0)
km = [1, 2, 3, 4, 5, 6, 7, 8]
minutes = [14, 15, 20, 21, 26, 27, 31, 35]
print(model.predict(4))
print(model.loss(km, minutes))
model.per_km = 3.0                # turn a knob
model.start = 10.0
print(model.loss(km, minutes))
````

### Arrays and their shape

An **array** is NumPy’s block of numbers, all of one type. You usually make one from a list:

````python
import numpy as np

km = np.array([1, 2, 3, 4, 5, 6, 7, 8])
minutes = np.array([14, 15, 20, 21, 26, 27, 31, 35])

print(km.shape)        # one dimension, 8 numbers
print(km.dtype)        # every item is the same type
print(km * 2)          # arithmetic on every item at once
print(km[2:5])         # slicing works like lists

grid = np.array([[1, 14], [2, 15], [3, 20]])
print(grid.shape)      # 3 rows, 2 columns
print(grid[:, 1])      # every row, column 1
````

### Arithmetic on whole arrays

In lesson 0.5 you worked out the pizza model’s misses by hand, one delivery at a time, and in lesson 2.2 with a comprehension. In NumPy the whole calculation is two lines, with no loop:

````python
import numpy as np

km = np.array([1, 2, 3, 4, 5, 6, 7, 8])
minutes = np.array([14, 15, 20, 21, 26, 27, 31, 35])

per_km, start = 2.0, 15.0
predicted = per_km * km + start            # all 8 at once
misses = np.abs(predicted - minutes)
print(predicted)
print(misses)
print(misses.mean())                       # the loss
````

### Broadcasting: when shapes differ

In `per_km * km`, a single number met an array of 8, and NumPy simply used that number for every item. That stretching has a name, **broadcasting**, and it also works between arrays. Here, each column of a small table has its own average subtracted:

````python
import numpy as np

# 3 deliveries, 2 columns: distance in km, minutes taken
data = np.array([[1.0, 14.0],
                 [4.0, 21.0],
                 [7.0, 31.0]])

col_means = data.mean(axis=0)   # average down each column
print(col_means)
print(col_means.shape)
print(data - col_means)         # (3, 2) minus (2,) -> (3, 2)
````

### See the speed yourself

Here is the race from the start of the lesson. It uses a seeded random generator, so the numbers are the same each time; `1_000_000` is a million, with underscores to make it easier to read:

````python
import time
import numpy as np

rng = np.random.default_rng(0)       # seeded random numbers
a = rng.random(1_000_000)
b = rng.random(1_000_000)

def loop_dot(xs, ys):
    total = 0.0
    for x, y in zip(xs, ys):         # one pair at a time
        total += x * y
    return total

a_list, b_list = a.tolist(), b.tolist()
t = time.perf_counter()
slow = loop_dot(a_list, b_list)
loop_ms = (time.perf_counter() - t) * 1000

t = time.perf_counter()
fast = np.dot(a, b)                  # all pairs in one call
numpy_ms = (time.perf_counter() - t) * 1000

print(f"loop {loop_ms:.1f} ms, NumPy {numpy_ms:.2f} ms")
print(f"same answer: {np.isclose(slow, fast)}")
````

### Plot what you compute: Anscombe’s quartet

Here is the quartet from the start of the lesson, built as one DataFrame and drawn as four scatter plots, charts of dots, side by side:

A **checkpoint** is a saved snapshot of a job’s progress: everything it needs to carry on from that point, like a save point in a video game. This cell stands in for a long job of 1,000 steps:

````python
import json
import os

FOLDER = "/content/drive/MyDrive/course"
CKPT = FOLDER + "/count.json"
os.makedirs(FOLDER, exist_ok=True)   # make the folder if needed

if os.path.exists(CKPT):             # carry on from the last save
    with open(CKPT) as f:
        state = json.load(f)
else:                                # or start from scratch
    state = {"step": 0, "total": 0}
print("starting at step", state["step"])

for step in range(state["step"], 1000):
    state["total"] += step * step    # stand-in for one step of work
    state["step"] = step + 1
    if state["step"] % 100 == 0:     # every 100 steps,
        with open(CKPT, "w") as f:
            json.dump(state, f)      # save everything needed

print(state)
````
